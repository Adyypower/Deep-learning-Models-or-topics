```python
from transformers import AutoTokenizer
```

```python
messages = [{"role":"user","content":"how do chat templates work?"},
            {"role":"assistant","content":"Chat templates help LLms like me to genrate"},
             {"role":"user","content":"how do i use them?"}]
```

```python
tokenizer = AutoTokenizer.from_pretrained('facebook/blenderbot-400M-distill')
```

```python
tokenizer = AutoTokenizer.from_pretrained('unsloth/mistral-7b-instruct-v0.2')
```

```python
tokenizer.apply_chat_template(messages, tokenize=False,add_generation_prompt=True)
```
